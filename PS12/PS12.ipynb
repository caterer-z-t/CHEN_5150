Poisson processes. Single molecule detection is commonly used in this building to estimate kinetic rate processes. As one example, the Kaar and Schwartz groups (ref 1) labeled a protein with two fluorophores, and used FRET efficiency to monitor the transition between a folded state (F) and an unfolded state (U) assuming a two-state folding model.

$U \xrightarrow{k_f} \xleftarrow{k_u} F$

They determined the kinetic rate of unfolding, ku, by counting the number of times their protein was observed in a folded state and stayed in a folded state (Nff), and the number of times their protein was in a folded state and then unfolded in the next time period (Nfu). The equation correlating the kinetic rate of unfolding to these experimental observables are given by:

$k_u = - \frac{1}{\tau} log⁡ \left( \frac{N_{ff}}{N_{ff} + N_{fu}} \right) $

Consider the $N_fu$ to be governed by a Poisson process. $\tau$ is the time period between images, and is 80 ms (milliseconds) under the conditions of the paper.

# Problem 4

Stochastic simulation. Cell fate decisions by mammalian cells in response to viral infection can be stochastic – only 20% of cells will produce interferon beta. Another way of saying this is that every cell can produce interferon beta, given enough time, but at a given time point only a subset would be able do so. Literature (ref 2) suggests that stochasticity is governed by low mRNA copy numbers encoding proteins which sense viral DNA, proteins which activate the signaling pathway, to the transcription factors which promote expression of interferon beta. Let’s consider a simplified version of this pathway, which operates by positive feedback. Consider a protein P which senses viral DNA. Assume for simplicity that the amount of DNA is constant and in great excess over the number of proteins. The protein can interact with DNA, forming a complex C. The complex C can stimulate production of the protein P. Production of interferon beta occurs when the number of complexes reach 10 or more. Degradation of P and C occur using the same rate constant $\mu$ and are 1st order. There are many simplifications to the system – we’re not considering mRNA transcription and translation, we’re not considering other proteins in the pathway, we’re not considering the gene regulatory network for interferon beta gene expression, and so forth. The biochemical network is shown below. 

$ \emptyset \xrightarrow{\lambda1 + \lambda2 \left(\frac{C}{K + C}\right) } P; P \xrightarrow{k_1} C; C \xrightarrow{k_{off}} P; P \xrightarrow{\mu} O; C \xrightarrow{\mu} \emptyset $

# Problem 4a

Write the stoichiometric matrix and the associated rate vector for this process.


$\mathbf{x} = [P, C]^T$

Stoichiometric matrix: (rows = species, cols = reactions):

$S = \begin{bmatrix} +1 & -1 & +1 & -1 & 0 \\ 0 & +1 & -1 & 0 & -1 \end{bmatrix}$

Rate vector:

$\mathbf{r} = \begin{bmatrix} \lambda_1 + \lambda_2\dfrac{C}{K+C} \\ k_1 P \\ k_{off} C \\ \mu P \\ \mu C \end{bmatrix}$

## Problem 4b
Simulate this stochastic network using initial amounts of P, C = 0. Show a plot of number of complexes C vs. time (in hours). Determine the amount of time it takes to reach 10 complexes.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters (units: h^-1)
lambda1 = 0.2
lambda2 = 20.0
k1      = 20.0
koff    = 10.0
mu      = 1.0
K       = 1.0   # half-saturation for feedback (molecules)

def gillespie(t_end=30.0, seed=None):
    rng = np.random.default_rng(seed)
    P, C = 0, 0
    t = 0.0
    ts, Cs = [0.0], [0]
    t_hit = None

    while t < t_end:
        r = np.array([
            lambda1 + lambda2 * C / (K + C),  # production of P
            k1 * P,                            # P -> C
            koff * C,                          # C -> P
            mu * P,                            # P degradation
            mu * C,                            # C degradation
        ])
        R = r.sum()
        if R == 0:
            break

        dt = rng.exponential(1.0 / R)
        t += dt

        u = rng.uniform(0, R)
        cumr = np.cumsum(r)
        if u < cumr[0]:
            P += 1
        elif u < cumr[1]:
            P -= 1; C += 1
        elif u < cumr[2]:
            C -= 1; P += 1
        elif u < cumr[3]:
            P -= 1
        else:
            C -= 1

        P = max(P, 0)
        C = max(C, 0)
        ts.append(t)
        Cs.append(C)

        if t_hit is None and C >= 10:
            t_hit = t

    return np.array(ts), np.array(Cs), t_hit


if __name__ == "__main__":
    ts, Cs, t_hit = gillespie(seed=42)

    print(f"Time to reach C >= 10: {t_hit:.4f} h" if t_hit else "C = 10 not reached")

    fig, ax = plt.subplots(figsize=(7, 4))
    ax.step(ts, Cs, where="post", color="#378ADD", linewidth=1.2, label="C (complexes)")
    ax.axhline(10, color="#E24B4A", linestyle="--", linewidth=1.0, label="C = 10 threshold")
    if t_hit:
        ax.axvline(t_hit, color="#E24B4A", linestyle=":", linewidth=0.8)
        # ax.annotate(f"t = {t_hit:.2f} h", xy=(t_hit, 10),
        #             xytext=(t_hit + 0.3, 11), fontsize=9, color="#E24B4A")
    ax.set_xlabel("Time (h)")
    ax.set_ylabel("Number of complexes C")
    ax.set_title("Interferon-β pathway")
    ax.legend(fontsize=9)
    plt.tight_layout()
    plt.show()
```

## Problem 4c

Repeat the simulation in part b 50 times. Plot the distribution of the amount of time it takes to reach 10 complexes. Determine the time in hours for 20% of the cells to be activated.

| Parameter | Value | Units |
|-----------|-------|-------|
| $\lambda 1$ | 0.2 | $h^{-1} $ |
| $\lambda 2$ | 20  | $h^{-1} $ |
| $k_1$       | 20  | $h^{-1} $ |
| $k_{off}$   | 10  | $h^{-1} $ |
| $\mu$.      | 1.  | $h^{-1} $ |

```python
import numpy as np
import matplotlib.pyplot as plt

lambda1 = 0.2
lambda2 = 20.0
k1      = 20.0
koff    = 10.0
mu      = 1.0
K       = 1.0

def gillespie(t_end=30.0, seed=None):
    rng = np.random.default_rng(seed)
    P, C = 0, 0
    t = 0.0
    ts, Cs = [0.0], [0]
    t_hit = None
    while t < t_end:
        r = np.array([
            lambda1 + lambda2 * C / (K + C),
            k1 * P,
            koff * C,
            mu * P,
            mu * C,
        ])
        R = r.sum()
        if R == 0:
            break
        dt = rng.exponential(1.0 / R)
        t += dt
        u = rng.uniform(0, R)
        cumr = np.cumsum(r)
        if u < cumr[0]:
            P += 1
        elif u < cumr[1]:
            P -= 1; C += 1
        elif u < cumr[2]:
            C -= 1; P += 1
        elif u < cumr[3]:
            P -= 1
        else:
            C -= 1
        P = max(P, 0)
        C = max(C, 0)
        ts.append(t)
        Cs.append(C)
        if t_hit is None and C >= 10:
            t_hit = t
    return np.array(ts), np.array(Cs), t_hit


N = 50
hit_times = []
for i in range(N):
    _, _, t_hit = gillespie(t_end=30.0, seed=i)
    if t_hit is not None:
        hit_times.append(t_hit)

hit_times = np.array(hit_times)
t20 = np.percentile(hit_times, 20)

print(f"Runs that reached C>=10: {len(hit_times)}/{N}")
print(f"20th percentile activation time: {t20:.4f} h")

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(hit_times, bins=15, color="#378ADD", edgecolor="white", linewidth=0.5)
ax.axvline(t20, color="#E24B4A", linestyle="--", linewidth=1.5,
           label=f"20th percentile = {t20:.2f} h")
ax.set_xlabel("Time to reach C = 10 (h)")
ax.set_ylabel("Count")
ax.set_title("Distribution of activation times (50 simulations)")
ax.legend(fontsize=9)
plt.tight_layout()
plt.show()
```
